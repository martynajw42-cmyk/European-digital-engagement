### Dataset Description

The dataset selected for this project is *Europeans Using Online Public Services (2008–2025)*, obtained from Kaggle. It is based on official statistics published by Eurostat and examines the use of digital public services across European countries.

The dataset contains information on online interactions between individuals and public authorities, including obtaining information, submitting forms and using government websites or applications.

The variables include country, country code, year, type of online activity, demographic group and the percentage of individuals engaging in each activity. Demographic breakdowns include age, gender and educational attainment.

This dataset is relevant to the wider study of digital engagement and civic participation in Europe. It provides an opportunity to explore differences in online engagement across countries and socio-demographic groups.

### Initial Data Quality Assessment

The initial inspection identified 35,852 observations and 14 variables, covering 39 geographical entries and the period 2008–2025.

The dataset contains information on different forms of online interaction with public authorities, with demographic breakdowns covering age, gender and educational attainment.

Several data-quality considerations were identified:

1. **Missing values:** The variables `flag` and `flag_meaning` contain approximately 97.74% missing values. These variables provide additional information about data quality and should not be confused with missing observations in the main percentage variable, `share_pct`.

2. **Country identifiers:** The `country_iso2` variable contains approximately 2.95% missing values. These observations require further investigation to determine whether they represent aggregated geographical categories.

3. **Uneven temporal coverage:** The number of observations varies considerably between years, particularly in 2013 and from 2022 onwards. Changes in indicators and demographic coverage may contribute to these differences.

4. **Geographical coverage:** The dataset includes EU and non-EU European countries, as well as an aggregated EU-27 category. This distinction will be important when interpreting results in the context of European Union citizens.

5. **Duplicate observations:** No completely duplicated rows were identified during the initial inspection.

Further investigation is required before undertaking data cleaning and merging.

```python
import requests
import pandas as pd

url = "https://ec.europa.eu/eurostat/api/dissemination/statistics/1.0/data/isoc_ci_ac_i"

params = {
    "lang": "en",
    "indic_is": "I_IUNW1",
    "unit": "PC_IND",
    "ind_type": "Y16_24"
}

response = requests.get(url, params=params)
response.raise_for_status()
data = response.json()

# Convert Eurostat JSON-stat data into a DataFrame
from pyjstat import pyjstat
```

```python
import json
from pyjstat import pyjstat

df_news = pyjstat.Dataset.read(json.dumps(data)).write('dataframe')

df_news.head()
```

```python
print(df_news.columns.tolist())
print(df_news['Geopolitical entity (reporting)'].unique()[:15])
print(df_news['value'].notna().sum())
```

```python
# Keep country-level observations with actual values
df_news = df_news[
    df_news['value'].notna()
].copy()

eu_groups = df_news['Geopolitical entity (reporting)'].str.startswith(
    ('European Union', 'Euro area')
)

df_news = df_news[~eu_groups].copy()

# Rename columns for easier merging
df_news = df_news.rename(columns={
    'Geopolitical entity (reporting)': 'country',
    'Time': 'year',
    'value': 'news_pct'
})

df_news['year'] = df_news['year'].astype(int)
df_news['news_pct'] = pd.to_numeric(df_news['news_pct'])

df_news = df_news[['country', 'year', 'news_pct']]

# Save the complementary dataset
import os
os.makedirs('data/raw', exist_ok=True)
df_news.to_csv('data/raw/eurostat_online_news.csv', index=False)

print(df_news.shape)
df_news.head()
```

```python
# Keep only young people aged 16–24 in the Kaggle dataset
young_people = df[
    df['group'] == 'Individuals, 16 to 24 years old'
].copy()

# Keep the country-level records, excluding EU aggregates
young_people = young_people[
    young_people['is_country_group'] == False
].copy()

# Merge both datasets
merged = pd.merge(
    young_people,
    df_news,
    on=['country', 'year'],
    how='inner'
)

print("Merged shape:", merged.shape)
print("Countries:", merged['country'].nunique())
print("Years:", merged['year'].min(), "-", merged['year'].max())

merged.head()
```

```python
# Save merged dataset
import os
os.makedirs('data/processed', exist_ok=True)
merged.to_csv('data/processed/merged_dataset.csv', index=False)

# Visualise online news consumption by year
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10, 6))
sns.lineplot(data=df_news, x='year', y='news_pct', errorbar=None)
plt.title('Online News Consumption Among Young Europeans')
plt.xlabel('Year')
plt.ylabel('Percentage of individuals aged 16–24')
plt.tight_layout()
plt.show()
```

```python
# Plot 2: Public-service engagement by indicator
plt.figure(figsize=(12, 6))
sns.barplot(
    data=merged,
    x='share_pct',
    y='indicator',
    errorbar=None
)
plt.title('Online Public-Service Engagement by Activity')
plt.xlabel('Percentage')
plt.ylabel('Activity')
plt.tight_layout()
plt.show()
```

```python
# Average public-service engagement by country and year
country_year = merged.groupby(
    ['country', 'year', 'news_pct'],
    as_index=False
)['share_pct'].mean()

plt.figure(figsize=(9, 6))

sns.regplot(
    data=country_year,
    x='news_pct',
    y='share_pct',
    scatter_kws={'alpha': 0.4},
    line_kws={'color': 'red'}
)

plt.title('Online News Consumption vs Public-Service Engagement')
plt.xlabel('Online news consumption (%)')
plt.ylabel('Average public-service engagement (%)')
plt.tight_layout()
plt.show()
```

```python
print("Average news consumption:", round(country_year['news_pct'].mean(), 2))
print("Average public-service engagement:", round(country_year['share_pct'].mean(), 2))
print("Correlation:", round(country_year['news_pct'].corr(country_year['share_pct']), 3))
print("Countries:", country_year['country'].nunique())
print("Years:", country_year['year'].min(), "-", country_year['year'].max())
```

## Findings
The merged dataset contained observations from 37 European countries between 2013 and 2025. Average online news consumption among individuals aged 16–24 was 71.77%, while average engagement with online public services was 39.66%.

The correlation analysis produced a Pearson correlation coefficient of r = 0.515, indicating a moderate positive association between online news consumption and online public-service engagement. This suggests that countries and years with higher levels of online news consumption tended also to show higher levels of engagement with online public services.

However, this relationship does not establish causation. Other factors, including digital literacy, education, internet access and national differences in digital infrastructure, may also influence online engagement.
Limitations
Include these points in your report:
1. Different data sources: The Kaggle dataset is based on Eurostat statistics, while the complementary dataset was retrieved directly from the Eurostat API.
2. Different measures: Reading online news and using online public services measure different forms of digital engagement.
3. Aggregated data: The analysis uses country-level percentages rather than individual respondents. Results cannot be used to infer individual behaviour.
4. Missing observations: Not every country has data for every year.
5. Repeated values: News-consumption percentages are repeated across multiple public-service indicators in the merged dataset.
6. Correlation does not imply causation: The observed association does not demonstrate that news consumption causes greater public-service engagement.
